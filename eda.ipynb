```python
import ast

def convert_genres(text):
    return[genre['name'] for genre in ast.literal_eval(text)]
df['genres'] = df['genres'].apply(convert_genres)    
```

```python
def convert_keywords(text):
    return[keyword['name'] for keyword in ast.literal_eval(text)]
df['keywords'] = df['keywords'].apply(convert_keywords)  
     
```

```python
df['cast'].iloc[0]
```

```python
def convert_cast(text):
    return[cast['name'] for cast in ast.literal_eval(text)[:3]]
df['cast'] = df['cast'].apply(convert_cast)   
```

```python
df
```

```python
def convert_crew(text):
    for crew in ast.literal_eval(text):
        if crew['job'] == 'Director':
            return crew['name']
df['crew'] = df['crew'].apply(convert_crew)
```

```python
df['tags'] = (
    df['genres'].apply(lambda x: ' '.join(x)) + ' ' +
    df['keywords'].apply(lambda x: ' '.join(x)) + ' ' +
    df['overview'] + ' ' +
    df['cast'].apply(lambda x: ' '.join(x)) + ' ' +
    df['crew']
)
```
